```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.linspace(0,5,11)
y=x**2
```

```python
plt.plot(x,y)
plt.title("Hello")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
```

```python
plt.subplot(2,2,1)
plt.plot(x,x)
plt.title("Y=X")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

plt.subplot(2,2,2)
plt.plot(x,2*x)
plt.title("Y=2X")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.subplot(2,2,3)
plt.plot(x,y)
plt.title("Y=X^2")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

plt.subplot(2,2,4)
plt.plot(x**x,x)
plt.title("Y=root(X)")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
```

```python
fig=plt.figure(figsize=(20,8),dpi = 1000)
axis2=fig.add_axes([0.1,0.1,1,1])
axis2.plot(x,y)
axis1=fig.add_axes([0.3,0.4,0.5,0.5])
axis1.plot(x,y)
```
