## Question 1
**Rename one column**

**Task:** Rename 'Name' to 'Student_Name'.

```python
import pandas as pd

df=pd.read_csv("students.csv")

# ✍️ Write your solution below
df=df.rename(columns={"Name":"Student_Name"})
```

Uses DataFrame.rename() to rename a single column.
